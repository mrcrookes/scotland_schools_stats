## Setup libraries and functions

```python
# Import key librarys
import pandas as pd
import numpy as np
```

```python
# Define function to read in excel file
def readfile(year, sheet, filetype, rows):
    
        df = pd.read_excel('./attendance/' + year + filetype, 
                                sheet_name=sheet,
                                skiprows=rows,
                                na_values = ['z', 'c', ',c', 'C','x', '#', '*', '.']
                                )
        return df
    
# Define tidy up initial df
def tidy(df):
    
    #### Fix usual issues with all strings
    
    # Capitalise columns
    df = df.map(lambda x: x.upper() if type(x) is str else x)

    # Strip whitespace
    df = df.map(lambda x: x.strip() if type(x) is str else x)

    # Remove parenthesis
    df = df.map(lambda x: x.replace('(', '') if type(x) is str else x)
    df = df.map(lambda x: x.replace(')', '') if type(x) is str else x)
    
    # Remove linebreaks
    df = df.map(lambda x: x.replace('\n', '') if type(x) is str else x)

    # Replace annoying substrings
    df = df.map(lambda x: x.replace(' AND ', ' & ') if type(x) is str else x)
    df = df.map(lambda x: x.replace(' – ', ' - ') if type(x) is str else x)
    df = df.map(lambda x: x.replace(' / ', '/') if type(x) is str else x)
    df = df.map(lambda x: x.replace('/ ', '/') if type(x) is str else x)
    df = df.map(lambda x: x.replace(' /', '/') if type(x) is str else x)
    
    # Remove suffixes
    for i in range(8):
        
        df = df.map(lambda x: x.removesuffix(" [NOTE " + str(i) + "]") if type(x) is str else x)
        df = df.map(lambda x: x.removesuffix("[NOTE " + str(i) + "]") if type(x) is str else x)
        df = df.map(lambda x: x.removesuffix(" " + str(i)) if type(x) is str else x)
        df = df.map(lambda x: x.removesuffix(str(i)) if type(x) is str else x)
    
    # Capitalise headers
    df.columns = df.columns.astype(str).str.upper()

    # Remove parenthesis from column headers
    df.columns = df.columns.map(lambda x: x.replace('(', '') if type(x) is str else x)
    df.columns = df.columns.map(lambda x: x.replace(')', '') if type(x) is str else x)

    # Remove suffixes from column headers

    for i in reversed(range(10)):

        df.columns = df.columns.map(lambda x: x.removesuffix("[NOTE " + str(i) + "]") if type(x) is str else x)

    # Remove remain [Note 8], I think not removed as next to %
    df.columns = df.columns.str.replace('[NOTE 8]', '', regex=False)
        
    # Strip whitespace from headers
    df.columns = df.columns.map(lambda x: x.strip() if type(x) is str else x)

    return df
```

## Read in data

```python
# Create frames to store ldfs
frames = []
```

```python
# Read in more recent data

# Create list of years
years = ['2023', '2024']

# Loop through years
for y in years:
    
    # Read in df
    wdf = readfile(y, 'Table 3.2', '.xlsx', 1)
    
    # Tidy df
    wdf = tidy(wdf)
    
    # Overwrite original column with percentage
    wdf['ATTENDANCE'] = wdf['ATTENDANCE'] / wdf['POSSIBLE ATTENDANCE'] * 100
    
    # Rename LA column
    wdf = wdf.rename(columns={'LOCAL AUTHORITY NAME': 'LEA', 'SCHOOL NAME': 'SCHOOL'})
    
    # Select best columns
    wdf = wdf[['SEEDCODE', 'SCHOOL', 'LEA', 'ATTENDANCE']]
    
#     # Overwrite original column with percentage
#     wdf['AUTHORISED ABSENCE'] = wdf['AUTHORISED ABSENCE'] / wdf['POSSIBLE ATTENDANCE'] * 100
    
#     # Overwrite original column with percentage
#     wdf['UNAUTHORISED ABSENCE'] = wdf['UNAUTHORISED ABSENCE'] / wdf['POSSIBLE ATTENDANCE'] * 100
    
#     # Overwrite original column with percentage
#     wdf['TEMPORARY EXCLUSION'] = wdf['TEMPORARY EXCLUSION'] / wdf['POSSIBLE ATTENDANCE'] * 100
    
    # Melt data into long format
    ldf = pd.melt(wdf, id_vars=['SEEDCODE', 'SCHOOL', 'LEA'],
                    value_vars=wdf.columns, var_name='VARIABLE', value_name='VALUE')
    
    # Add column for year
    ldf['YEAR'] = y
        
    # Append to list of df (with year)
    frames.append(ldf)
```

## Join, tidy and export data

```python
# Concat all ldf together
tdf = pd.concat(frames)

# # Replace zeros (for attendance I think these are schools that have just opened) with NaN
# tdf.replace(0, np.nan, inplace=True)

# Drop any rows with nans
tdfc = tdf.dropna().copy()

# Convert SEEDCODE column to in
tdfc['SEEDCODE'] = tdfc['SEEDCODE'].apply(lambda x: int(x))

## Clean up school names using names from 2024 & 2015 & 2010

# Select only 2024 data (before making any changes!)
df2024 = tdfc.loc[tdfc['YEAR'] == '2024']

# Select only 2015 data (before making any changes!)
df2015 = tdfc.loc[tdfc['YEAR'] == '2015']

# Select only 2010 data (before making any changes!)
df2010 = tdfc.loc[tdfc['YEAR'] == '2010']

# Create dictionary of seedcode : school pairs
skey = dict(zip(df2010.SEEDCODE,df2010.SCHOOL))
# Create new column with 2010 values (fill any blank values with original values)
tdfc['SCHOOL'] = tdfc['SEEDCODE'].map(skey).fillna(tdfc['SCHOOL'])

# Create dictionary of seedcode : school pairs
skey = dict(zip(df2015.SEEDCODE,df2015.SCHOOL))
# Create new column with 2015 values (fill any blank values with original values)
tdfc['SCHOOL'] = tdfc['SEEDCODE'].map(skey).fillna(tdfc['SCHOOL'])

# Create dictionary of seedcode : school pairs
skey = dict(zip(df2024.SEEDCODE,df2024.SCHOOL))
# Create new column with 2024 values (fill any blank values with original values)
tdfc['SCHOOL'] = tdfc['SEEDCODE'].map(skey).fillna(tdfc['SCHOOL'])

# Clean up remaining school name
tdfc['SCHOOL'].replace('GLASGOW GAELIC SECONDARY SC', 'GLASGOW GAELIC SECONDARY SCHOOL', inplace=True)

## Clean up councils using names from 2024 & 2010

# Create dictionary of seedcode : council pairs
lkey = dict(zip(df2010.SEEDCODE,df2010.LEA))
# Create new column with 2010 values (fill any blank values with original values)
tdfc['LEA'] = tdfc['SEEDCODE'].map(lkey).fillna(tdfc['LEA'])

# Create dictionary of seedcode : council pairs
lkey = dict(zip(df2024.SEEDCODE,df2024.LEA))
# Create new column with 2024 values (fill any blank values with original values)
tdfc['LEA'] = tdfc['SEEDCODE'].map(lkey).fillna(tdfc['LEA'])

# Sort values by seedcode
tdfc.sort_values('SEEDCODE', inplace=True)

# Clean up remaining council name
tdfc['LEA'].replace('WEST DUNBARTONS', 'WEST DUNBARTONSHIRE', inplace=True)
```
